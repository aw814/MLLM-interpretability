```python
import pandas as pd
import os


folder_name = '/Users/anniewang/Desktop/MLLM-interpretability/MLLM-interpretability/eval/artifacts/gpt-4o'  # replace with your folder path
for file in os.listdir(folder_name):
    if 'predictions.csv' in file:
        print(file)
        merged_df = pd.read_csv(os.path.join(folder_name, file))
        tgt_lang = merged_df['target_lang'][0]
        # long_dataset = pd.read_csv("/Users/anniewang/Desktop/MLLM-interpretability/MLLM-interpretability/data/processed/eclektic_long_subset.csv")
        # long_dataset = long_dataset[long_dataset['language'] == tgt_lang]
        # merged_df = predicition.merge(long_dataset, on='q_id', how='left')
        merged_df = merged_df.drop(columns=["original_question","question", "language"])
        merged_df = merged_df.rename(columns={
                                        'original_content': 'c_src',
                                        'content': 'c_tgt',
                                        "original_answer": 'gold_answer_src',
                                        "answer": 'gold_answer_tgt'
                                    })
        output_dir = folder_name
        output_path = os.path.join(output_dir, f"{tgt_lang}_predictions.csv")
        merged_df.to_csv(output_path, index=False)

        
```
