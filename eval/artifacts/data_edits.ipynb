```python
import pandas as pd
import os


folder_name = '/Users/anniewang/Desktop/MLLM-interpretability/MLLM-interpretability/eval/artifacts/gemini-2.0-flash'  # replace with your folder path
for file in os.listdir(folder_name):
    if 'predictions.csv' in file:
        print(file)
        predicition = pd.read_csv(os.path.join(folder_name, file))
        tgt_lang = predicition['target_lang'][0]
        long_dataset = pd.read_csv("/Users/anniewang/Desktop/MLLM-interpretability/MLLM-interpretability/data/processed/eclektic_long_subset.csv")
        long_dataset = long_dataset[long_dataset['language'] == tgt_lang]
        merged_df = predicition.merge(long_dataset, on='q_id', how='left')
        output_dir = os.path.join(folder_name, 'new')
        os.makedirs(output_dir, exist_ok=True)
        output_path = os.path.join(output_dir, f"{tgt_lang}_predictions.csv")
        merged_df.to_csv(output_path, index=False)

        
```
