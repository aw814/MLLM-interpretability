```python
import pandas as pd

imp = pd.read_csv("model_feature_importances.csv")

# Top 20 features for LogisticRegression using all features
imp[(imp["feature_set"] == "all") & (imp["model"] == "LogisticRegression")] \
   .groupby("orig_feature_set")["abs_importance"].max() \
   .sort_values(ascending=False).head(20)

```
