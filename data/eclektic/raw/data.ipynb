```python
import os
import pandas as pd
import json

dataset_name = "eclektic"
```

```python
df = pd.read_csv(f"{dataset_name}_7.csv")

# generate a new column '{dataset_name}_id' by concatenating 'language' string,'q_id' int64
df[f"{dataset_name}_id"] = df["language"] + "_" + df["q_id"].astype(str)
```

```python
df.head()
```

```python
with open(f"{dataset_name}_batch_requests.jsonl", "w") as f:
    for _, row in df.iterrows():

        request = {
            "key": row["eclektic_id"],
            "request": {
                "contents": [
                    {
                        "role": "user",
                        "parts": [{"text": row["question"]}]
                    }
                ],
                "generation_config": {
                    "temperature": 0,
                    "top_p": 1,
                    "max_output_tokens": 64
                }
            }
        }

        f.write(json.dumps(request, ensure_ascii=False) + "\n")
```
